```r
summary(attgt)

# Event-study dinámico
es <- aggte(attgt, type = "dynamic", balance_e = TRUE)

# Tabla de coeficientes
event_table <- data.frame(
  k = es$egt,
  ATT = es$att,
  se = es$se,
  ci_lower = es$att - 1.96*es$se,
  ci_upper = es$att + 1.96*es$se
)

print(event_table)
```

```r
agg_group <- aggte(attgt, type = "group")

# Tabla limpia
group_table <- data.frame(
  cohort = agg_group$g,
  ATT = agg_group$att,
  se = agg_group$se,
  ci_lower = agg_group$att - 1.96*agg_group$se,
  ci_upper = agg_group$att + 1.96*agg_group$se
)

print(group_table)
```

```r
agg_period <- aggte(attgt, type = "calendar")

# Tabla limpia
period_table <- data.frame(
  year = agg_period$egt,   # años de observación
  ATT = agg_period$att,
  se = agg_period$se,
  ci_lower = agg_period$att - 1.96*agg_period$se,
  ci_upper = agg_period$att + 1.96*agg_period$se
)

print(period_table)
```

```r
agg_event <- aggte(attgt, type = "dynamic", balance_e = TRUE)

# Tabla limpia
event_table <- data.frame(
  event_time = agg_event$egt,  # tiempo relativo k
  ATT = agg_event$att,
  se = agg_event$se,
  ci_lower = agg_event$att - 1.96*agg_event$se,
  ci_upper = agg_event$att + 1.96*agg_event$se
)

print(event_table)
```
